```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class CNNBlock(nn.Module):
    """Stacked 1D convolutions extract local waveform features."""
    def __init__(self, in_channels=12, hidden=64, out_channels=128, dropout=0.2):
        super().__init__()
        self.conv1 = nn.Conv1d(in_channels, hidden, kernel_size=7, padding=3)
        self.bn1 = nn.BatchNorm1d(hidden)
        self.conv2 = nn.Conv1d(hidden, hidden, kernel_size=5, padding=2)
        self.bn2 = nn.BatchNorm1d(hidden)
        self.conv3 = nn.Conv1d(hidden, out_channels, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm1d(out_channels)
        self.pool = nn.MaxPool1d(kernel_size=2)
        self.drop = nn.Dropout(dropout)

    def forward(self, x):
        x = self.pool(F.relu(self.bn1(self.conv1(x))))
        x = self.pool(F.relu(self.bn2(self.conv2(x))))
        x = self.pool(F.relu(self.bn3(self.conv3(x))))
        return self.drop(x)


class AutoencoderBlock(nn.Module):
    """Compresses CNN features to a latent z; reconstruction loss is an aux regularizer."""
    def __init__(self, in_channels=128, latent=64):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Conv1d(in_channels, 96, kernel_size=3, padding=1), nn.ReLU(),
            nn.Conv1d(96, latent, kernel_size=3, padding=1),
        )
        self.decoder = nn.Sequential(
            nn.Conv1d(latent, 96, kernel_size=3, padding=1), nn.ReLU(),
            nn.Conv1d(96, in_channels, kernel_size=3, padding=1),
        )

    def forward(self, x):
        z = self.encoder(x)
        x_hat = self.decoder(z)
        return z, x_hat


class BiLSTMBlock(nn.Module):
    def __init__(self, input_size, hidden=64, num_layers=1, dropout=0.2):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden, num_layers=num_layers,
                            batch_first=True, bidirectional=True,
                            dropout=dropout if num_layers > 1 else 0.0)

    def forward(self, x):
        out, _ = self.lstm(x)
        return out


class GRUBlock(nn.Module):
    def __init__(self, input_size, hidden=64, num_layers=1, dropout=0.2):
        super().__init__()
        self.gru = nn.GRU(input_size, hidden, num_layers=num_layers,
                          batch_first=True,
                          dropout=dropout if num_layers > 1 else 0.0)

    def forward(self, x):
        out, _ = self.gru(x)
        return out


class TemporalAttention(nn.Module):
    def __init__(self, hidden):
        super().__init__()
        self.attn = nn.Linear(hidden, 1)

    def forward(self, x):
        scores = self.attn(x).squeeze(-1)
        weights = torch.softmax(scores, dim=1)
        context = torch.sum(x * weights.unsqueeze(-1), dim=1)
        return context, weights


class ECGNet(nn.Module):
    """Full 5-block model. Toggle blocks off for ablation."""
    def __init__(self, in_channels=12, num_classes=5,
                 use_cnn=True, use_ae=True,
                 use_bilstm=True, use_gru=True, use_attention=True,
                 dropout=0.3):
        super().__init__()
        self.flags = dict(cnn=use_cnn, ae=use_ae, bilstm=use_bilstm,
                          gru=use_gru, attn=use_attention)
        c = in_channels
        self.cnn = CNNBlock(c, out_channels=128) if use_cnn else None
        if use_cnn: c = 128
        self.ae = AutoencoderBlock(c, latent=64) if use_ae else None
        if use_ae: c = 64
        self.bilstm = BiLSTMBlock(c, hidden=64) if use_bilstm else None
        if use_bilstm: c = 128
        self.gru = GRUBlock(c, hidden=64) if use_gru else None
        if use_gru: c = 64
        self.attention = TemporalAttention(c) if use_attention else None
        self.classifier = nn.Sequential(
            nn.Linear(c, 64), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(64, num_classes),
        )

    def forward(self, x):
        recon_target = recon_pred = None
        if self.flags['cnn']:
            x = self.cnn(x)
        if self.flags['ae']:
            recon_target = x
            x, recon_pred = self.ae(x)
        x = x.transpose(1, 2)
        if self.flags['bilstm']:
            x = self.bilstm(x)
        if self.flags['gru']:
            x = self.gru(x)
        if self.flags['attn']:
            x, _ = self.attention(x)
        else:
            x = x.mean(dim=1)
        return self.classifier(x), recon_target, recon_pred


def compute_loss(logits, targets, recon_target, recon_pred,
                 ce_weight=1.0, recon_weight=0.1):
    ce = F.cross_entropy(logits, targets)
    if recon_target is not None and recon_pred is not None:
        recon = F.mse_loss(recon_pred, recon_target)
        return ce_weight * ce + recon_weight * recon, ce.item(), recon.item()
    return ce_weight * ce, ce.item(), 0.0


print("Model defined")
```

```python
import os
import ast
import numpy as np
import pandas as pd
import wfdb
from torch.utils.data import Dataset, DataLoader

SUPERCLASSES = ['NORM', 'MI', 'STTC', 'CD', 'HYP']
CLASS_TO_IDX = {c: i for i, c in enumerate(SUPERCLASSES)}


def _aggregate_superclasses(scp_codes, agg_df):
    out = set()
    for code in scp_codes.keys():
        if code in agg_df.index:
            sc = agg_df.loc[code, 'diagnostic_class']
            if isinstance(sc, str):
                out.add(sc)
    return list(out)


class PTBXLDataset(Dataset):
    """Caches the full dataset as .npy on first load -> very fast epochs after."""
    def __init__(self, data_root, sampling_rate=100, split='train',
                 cache_dir='/content/ptbxl_cache'):
        os.makedirs(cache_dir, exist_ok=True)
        sig_cache = os.path.join(cache_dir, f'{split}_signals_{sampling_rate}.npy')
        lab_cache = os.path.join(cache_dir, f'{split}_labels_{sampling_rate}.npy')

        if os.path.exists(sig_cache) and os.path.exists(lab_cache):
            self.signals = np.load(sig_cache)
            self.labels = np.load(lab_cache)
            print(f"[PTBXL/{split}] loaded {len(self.labels)} cached records")
            return

        df = pd.read_csv(os.path.join(data_root, 'ptbxl_database.csv'), index_col='ecg_id')
        df.scp_codes = df.scp_codes.apply(ast.literal_eval)
        agg = pd.read_csv(os.path.join(data_root, 'scp_statements.csv'), index_col=0)
        agg = agg[agg.diagnostic == 1]
        df['superclass'] = df.scp_codes.apply(lambda x: _aggregate_superclasses(x, agg))
        df = df[df.superclass.apply(len) == 1].copy()
        df['label'] = df.superclass.apply(lambda x: CLASS_TO_IDX[x[0]])

        if split == 'train':
            df = df[df.strat_fold.isin(range(1, 9))]
        elif split == 'val':
            df = df[df.strat_fold == 9]
        elif split == 'test':
            df = df[df.strat_fold == 10]

        fn_col = 'filename_lr' if sampling_rate == 100 else 'filename_hr'
        df = df.reset_index()
        signals, labels = [], []
        print(f"[PTBXL/{split}] caching {len(df)} records...")
        for i, row in df.iterrows():
            sig, _ = wfdb.rdsamp(os.path.join(data_root, row[fn_col]))
            sig = sig.T.astype(np.float32)
            sig = (sig - sig.mean(axis=1, keepdims=True)) / (sig.std(axis=1, keepdims=True) + 1e-8)
            signals.append(sig)
            labels.append(int(row['label']))
            if (i + 1) % 1000 == 0:
                print(f"  ...{i + 1}/{len(df)}")
        self.signals = np.stack(signals).astype(np.float32)
        self.labels = np.asarray(labels, dtype=np.int64)
        np.save(sig_cache, self.signals)
        np.save(lab_cache, self.labels)
        print(f"[PTBXL/{split}] cached {len(self.labels)} records")

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return torch.from_numpy(self.signals[idx]), int(self.labels[idx])


def get_dataloaders(data_root, sampling_rate=100, batch_size=64,
                    cache_dir='/content/ptbxl_cache'):
    train_ds = PTBXLDataset(data_root, sampling_rate, 'train', cache_dir=cache_dir)
    val_ds = PTBXLDataset(data_root, sampling_rate, 'val', cache_dir=cache_dir)
    test_ds = PTBXLDataset(data_root, sampling_rate, 'test', cache_dir=cache_dir)
    return (
        DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True),
        DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True),
        DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True),
    )


print("Data utilities defined")
```

```python
import time
import json
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from torch.optim.lr_scheduler import CosineAnnealingLR


def train_one_epoch(model, loader, optimizer, device, ce_weight=1.0, recon_weight=0.1):
    model.train()
    tot_loss = tot_ce = tot_rec = 0.0
    n = 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        logits, rt, rp = model(x)
        loss, ce, rec = compute_loss(logits, y, rt, rp, ce_weight, recon_weight)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()
        bs = x.size(0)
        tot_loss += loss.item() * bs
        tot_ce += ce * bs
        tot_rec += rec * bs
        n += bs
    return tot_loss / n, tot_ce / n, tot_rec / n


@torch.no_grad()
def evaluate(model, loader, device):
    model.eval()
    all_logits, all_labels = [], []
    for x, y in loader:
        x = x.to(device)
        logits, _, _ = model(x)
        all_logits.append(logits.cpu())
        all_labels.append(y)
    logits = torch.cat(all_logits)
    labels = torch.cat(all_labels).numpy()
    preds = logits.argmax(dim=1).numpy()
    probs = F.softmax(logits, dim=1).numpy()
    acc = accuracy_score(labels, preds)
    f1 = f1_score(labels, preds, average='macro')
    try:
        auc = roc_auc_score(labels, probs, multi_class='ovr', average='macro')
    except ValueError:
        auc = float('nan')
    return acc, f1, auc


def train_model(model, train_loader, val_loader, device,
                epochs=15, lr=1e-3, weight_decay=1e-4,
                ce_weight=1.0, recon_weight=0.1,
                ckpt_path='best_model.pt', verbose=True):
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = CosineAnnealingLR(optimizer, T_max=epochs)
    best_f1 = -1.0
    history = []
    for epoch in range(1, epochs + 1):
        t0 = time.time()
        train_loss, ce, rec = train_one_epoch(model, train_loader, optimizer, device, ce_weight, recon_weight)
        val_acc, val_f1, val_auc = evaluate(model, val_loader, device)
        scheduler.step()
        dt = time.time() - t0
        history.append(dict(epoch=epoch, train_loss=train_loss, ce=ce, recon=rec,
                            val_acc=val_acc, val_f1=val_f1, val_auc=val_auc, time=dt))
        if verbose:
            print(f"[{epoch:02d}/{epochs}] loss={train_loss:.4f} | "
                  f"val_acc={val_acc:.4f} val_f1={val_f1:.4f} val_auc={val_auc:.4f} | {dt:.1f}s")
        if val_f1 > best_f1:
            best_f1 = val_f1
            torch.save({'model_state': model.state_dict(), 'epoch': epoch, 'val_f1': val_f1}, ckpt_path)
    with open(ckpt_path.replace('.pt', '_history.json'), 'w') as f:
        json.dump(history, f, indent=2)
    return history, best_f1


print("Training utilities defined")
```

```python
def train_model(model, train_loader, val_loader, device,
                epochs=30, lr=1e-3, weight_decay=1e-4,
                ce_weight=1.0, recon_weight=0.1,
                ckpt_path='best_model.pt', verbose=True,
                patience=7):
    """Early-stopping variant: stops when val_f1 hasn't improved for `patience` epochs."""
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = CosineAnnealingLR(optimizer, T_max=epochs)
    best_f1 = -1.0
    epochs_no_improve = 0
    history = []

    for epoch in range(1, epochs + 1):
        t0 = time.time()
        train_loss, ce, rec = train_one_epoch(model, train_loader, optimizer, device, ce_weight, recon_weight)
        val_acc, val_f1, val_auc = evaluate(model, val_loader, device)
        scheduler.step()
        dt = time.time() - t0
        history.append(dict(epoch=epoch, train_loss=train_loss, ce=ce, recon=rec,
                            val_acc=val_acc, val_f1=val_f1, val_auc=val_auc, time=dt))
        if verbose:
            print(f"[{epoch:02d}/{epochs}] loss={train_loss:.4f} | "
                  f"val_acc={val_acc:.4f} val_f1={val_f1:.4f} val_auc={val_auc:.4f} | {dt:.1f}s")

        if val_f1 > best_f1:
            best_f1 = val_f1
            epochs_no_improve = 0
            torch.save({'model_state': model.state_dict(), 'epoch': epoch, 'val_f1': val_f1}, ckpt_path)
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"  ⏸ Early stopping at epoch {epoch}  (best val_f1={best_f1:.4f}, no improvement for {patience} epochs)")
                break

    with open(ckpt_path.replace('.pt', '_history.json'), 'w') as f:
        json.dump(history, f, indent=2)
    return history, best_f1

print("Early stopping enabled (patience=7)")
```

```python
DATA_ROOT = '/content/ptbxl/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.1'

train_loader, val_loader, test_loader = get_dataloaders(DATA_ROOT, sampling_rate=100, batch_size=64)

x, y = next(iter(train_loader))
print("Batch shape:", x.shape, "Labels:", y[:5].tolist())
```

```python
device = torch.device('cuda')
model = ECGNet().to(device)
print(f"Params: {sum(p.numel() for p in model.parameters()):,}")

hist, best_f1 = train_model(model, train_loader, val_loader, device,
                            epochs=5, ckpt_path='/content/drive/MyDrive/smoke.pt')
print(f"\nSmoke test val_f1: {best_f1:.4f}")
```

```python
TUNE_RESULTS = []
configs = [
    (3e-4, 0.1), (1e-3, 0.1), (3e-3, 0.1),   # vary lr
    (1e-3, 0.0), (1e-3, 0.3),                # vary recon weight
]
for lr, rw in configs:
    print(f"\n--- lr={lr}  recon_weight={rw} ---")
    m = ECGNet().to(device)
    _, val_f1 = train_model(m, train_loader, val_loader, device,
                            epochs=5, lr=lr, recon_weight=rw,
                            ckpt_path=f'/content/tune.pt', verbose=False)
    print(f"   val_f1 = {val_f1:.4f}")
    TUNE_RESULTS.append({'lr': lr, 'recon_weight': rw, 'val_f1': val_f1})

tune_df = pd.DataFrame(TUNE_RESULTS)
print("\n=== Tuning results ===")
print(tune_df.to_string(index=False))
tune_df.to_csv('/content/drive/MyDrive/ecg_ablation/tuning.csv', index=False)

best = tune_df.iloc[tune_df['val_f1'].idxmax()]
BEST_LR     = float(best['lr'])
BEST_RECON  = float(best['recon_weight'])
print(f"\nBest config: lr={BEST_LR}  recon_weight={BEST_RECON}")
```

```python
ABLATIONS = [
    {'name': 'full',         'flags': {}},
    {'name': 'no_cnn',       'flags': {'use_cnn': False}},
    {'name': 'no_ae',        'flags': {'use_ae': False}},
    {'name': 'no_bilstm',    'flags': {'use_bilstm': False}},
    {'name': 'no_gru',       'flags': {'use_gru': False}},
    {'name': 'no_attention', 'flags': {'use_attention': False}},
]

OUT_DIR = '/content/drive/MyDrive/ecg_ablation'
os.makedirs(OUT_DIR, exist_ok=True)

results = []
for cfg in ABLATIONS:
    name, flags = cfg['name'], cfg['flags']
    print(f"\n{'='*60}\n {name}  {flags}\n{'='*60}")
    model = ECGNet(**flags).to(device)
    ckpt = os.path.join(OUT_DIR, f'{name}.pt')
    _, best_val_f1 = train_model(model, train_loader, val_loader, device,
                                 epochs=30, lr=BEST_LR, recon_weight=BEST_RECON,
                                 ckpt_path=ckpt)
    state = torch.load(ckpt)
    model.load_state_dict(state['model_state'])
    test_acc, test_f1, test_auc = evaluate(model, test_loader, device)
    results.append({'ablation': name, 'val_f1': best_val_f1,
                    'test_acc': test_acc, 'test_f1': test_f1, 'test_auc': test_auc,
                    'params': sum(p.numel() for p in model.parameters())})
    pd.DataFrame(results).to_csv(os.path.join(OUT_DIR, 'results.csv'), index=False)
    print(f"--> {name}: test_f1={test_f1:.4f}")

print("\n=== Final results ===")
print(pd.DataFrame(results).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import json
import os
import numpy as np
from sklearn.metrics import confusion_matrix

OUT_DIR = '/content/drive/MyDrive/ecg_ablation'
FIG_DIR = '/content/drive/MyDrive/ecg_figures'
os.makedirs(FIG_DIR, exist_ok=True)

# --- 1. Training curves for the full model ---
with open(os.path.join(OUT_DIR, 'full_history.json')) as f:
    hist = json.load(f)
epochs = [h['epoch'] for h in hist]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(epochs, [h['train_loss'] for h in hist], label='Total loss')
axes[0].plot(epochs, [h['ce'] for h in hist], label='Cross-entropy')
axes[0].plot(epochs, [h['recon'] for h in hist], label='Reconstruction')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss'); axes[0].legend()
axes[0].set_title('Training Loss (Full Model)')
axes[1].plot(epochs, [h['val_acc'] for h in hist], label='Accuracy')
axes[1].plot(epochs, [h['val_f1'] for h in hist], label='Macro-F1')
axes[1].plot(epochs, [h['val_auc'] for h in hist], label='Macro-AUC')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Score'); axes[1].legend()
axes[1].set_title('Validation Metrics (Full Model)')
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'training_curves.png'), dpi=150, bbox_inches='tight')
plt.show()

# --- 2. Ablation bar chart ---
df = pd.read_csv(os.path.join(OUT_DIR, 'results.csv'))
fig, ax = plt.subplots(figsize=(8, 5))
colors = ['steelblue' if a == 'full' else 'lightcoral' for a in df['ablation']]
bars = ax.bar(df['ablation'], df['test_f1'], color=colors, edgecolor='black')
full_f1 = df.loc[df.ablation == 'full', 'test_f1'].iloc[0]
ax.axhline(full_f1, color='steelblue', linestyle='--', alpha=0.6, label=f'Full model (F1={full_f1:.3f})')
for bar, val in zip(bars, df['test_f1']):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', fontsize=9)
ax.set_ylabel('Test Macro-F1'); ax.set_title('Ablation Study — Test F1 by Variant')
plt.xticks(rotation=20)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'ablation.png'), dpi=150, bbox_inches='tight')
plt.show()

# --- 3. Confusion matrix for the full model on test set ---
state = torch.load(os.path.join(OUT_DIR, 'full.pt'))
model = ECGNet().to(device)
model.load_state_dict(state['model_state'])
model.eval()
all_preds, all_labels = [], []
with torch.no_grad():
    for x, y in test_loader:
        x = x.to(device)
        logits, _, _ = model(x)
        all_preds.append(logits.argmax(dim=1).cpu().numpy())
        all_labels.append(y.numpy())
preds = np.concatenate(all_preds); labels = np.concatenate(all_labels)
cm = confusion_matrix(labels, preds)
cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm_norm, cmap='Blues', vmin=0, vmax=1)
ax.set_xticks(range(5)); ax.set_yticks(range(5))
ax.set_xticklabels(SUPERCLASSES); ax.set_yticklabels(SUPERCLASSES)
ax.set_xlabel('Predicted'); ax.set_ylabel('True')
ax.set_title('Confusion Matrix (Full Model, Test)')
for i in range(5):
    for j in range(5):
        ax.text(j, i, f'{cm_norm[i,j]:.2f}', ha='center', va='center',
                color='white' if cm_norm[i,j] > 0.5 else 'black', fontsize=9)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.savefig(os.path.join(FIG_DIR, 'confusion_matrix.png'), dpi=150, bbox_inches='tight')
plt.show()

print(f"\nAll figures saved to {FIG_DIR}")
```
